## Filters

```python
pd.set_option('display.max_rows', None)
print(df[['ebeam', 'x', 'Q2', 'z', 'thpq', 'hms_p']][(df['x']!=-999) & (df['target']=='Al')]
      .drop_duplicates()
      .sort_values(by=['z']))
```

```python
pd.set_option('display.max_rows', None)
print(df[['target', 'ebeam', 'x', 'Q2', 'z', 'thpq', 'hms_p', 'run_type', 'run']][(df['target']=='C') & (df['ebeam']==10.6716) & (df['z']==0.5) & (df['thpq']==-0.8) & (df['run_type']=="PI-SIDIS")]
      .drop_duplicates()
      .sort_values(by=['z']))
```

```python
pd.set_option('display.max_rows', None)
print(df[['target', 'ebeam', 'x', 'Q2', 'z', 'thpq', 'hms_p', 'run_type', 'run']][(df['target']=='C') & (df['ebeam']==10.6716) & (df['shms_p']==-3.632) & (df['shms_th']==7.51) & (df['run_type']=="PI-SIDIS")]
      .drop_duplicates()
      .sort_values(by=['z']))
```

## For SSA MC

```python
pd.set_option('display.max_rows', None)
df_mc = (df[['target', 'ebeam', 'hms_p', 'hms_th', 'shms_p', 'shms_th', 'x', 'Q2', 'z', 'thpq', 'run_type']][(df['target']!='Dummy') & (df['ebeam']==10.6716) & (df['run_type'].str.contains("PI")) & (df['hms_p']<0)]
            .drop_duplicates(subset=['target', 'ebeam', 'x', 'Q2', 'z', 'thpq', 'run_type'])      
            .sort_values(by=['z', 'target', 'thpq', 'run_type'], 
                        ascending=[True, True, True, True]))
len(df_mc)
df_mc.head(10)
```
